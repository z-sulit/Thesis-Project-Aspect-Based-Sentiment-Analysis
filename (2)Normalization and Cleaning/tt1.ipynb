```python
import pandas as pd                                                                                                                                                                                                                                        
from pathlib import Path                                                                                                                                                                                                                                   
                                                                                                                                                                                                                                                            
# Absolute path configuration                                                                                                                                                                                                                              
DATASET_DIR = Path(r"D:\Ateneo de Davao\(1) Thesis-Project-Aspect-Based-Sentiment-Analysis\Dataset")                                                                                                                                                       
COFFEE_DIR = DATASET_DIR / "Coffeeshops"                                                                                                                                                                                                                   
MATCHA_DIR = DATASET_DIR / "Matcha Shops"                                                                                                                                                                                                                  
                                                                                                                                                                                                                                                            
def concatenate_csvs(source_dir, output_file):                                                                                                                                                                                                             
    source_path = Path(source_dir)                                                                                                                                                                                                                         
    csv_files = sorted(source_path.glob("*.csv"))                                                                                                                                                                                                          
                                                                                                                                                                                                                                                            
    if not csv_files:                                                                                                                                                                                                                                      
        print(f"No CSV files found in: {source_path}")                                                                                                                                                                                                     
        return                                                                                                                                                                                                                                             
                                                                                                                                                                                                                                                            
    all_dfs = []                                                                                                                                                                                                                                           
    for f in csv_files:                                                                                                                                                                                                                                    
        try:                                                                                                                                                                                                                                               
            df = pd.read_csv(f)                                                                                                                                                                                                                            
            all_dfs.append(df)                                                                                                                                                                                                                             
        except Exception as e:                                                                                                                                                                                                                             
            print(f"Error reading {f.name}: {e}")                                                                                                                                                                                                          
                                                                                                                                                                                                                                                            
    if all_dfs:                                                                                                                                                                                                                                            
        master_df = pd.concat(all_dfs, ignore_index=True)                                                                                                                                                                                                  
        if 'rating' in master_df.columns:                                                                                                                                                                                                                  
            master_df['rating'] = pd.to_numeric(master_df['rating'], errors='coerce')                                                                                                                                                                      
        master_df.to_csv(output_file, index=False)                                                                                                                                                                                                         
        print(f"Successfully created: {output_file} ({len(master_df):,} rows)")                                                                                                                                                                            
    else:                                                                                                                                                                                                                                                  
        print("No DataFrames to concatenate.")                                                                                                                                                                                                             
                                                                                                                                                                                                                                                            
# Generate separate master CSVs                                                                                                                                                                                                                            
concatenate_csvs(COFFEE_DIR, DATASET_DIR / "coffeeshops_master.csv")                                                                                                                                                                                       
concatenate_csvs(MATCHA_DIR, DATASET_DIR / "matcha_shops_master.csv")            
```
