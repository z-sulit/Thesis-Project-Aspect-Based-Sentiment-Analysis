```python
import os                                                                                                                                                               
    from pathlib import Path                                                                                                                                                
                                                                                                                                                                            
    MATCHA_DIR = Path("../Dataset/Matcha Shops")                                                                                                                            
    COFFEE_DIR = Path("../Dataset/Coffeeshops")                                                                                                                             
                                                                                                                                                                            
    def clean_shop_name(path):                                                                                                                                              
        stem = path.stem                                                                                                                                                    
        if stem.endswith("_reviews"):                                                                                                                                       
            stem = stem[:-8]                                                                                                                                                
        return stem.replace("_", " ").strip()                                                                                                                               
                                                                                                                                                                            
    # Collect and clean shop names from CSV files                                                                                                                           
    coffee_shops = [clean_shop_name(p) for p in COFFEE_DIR.glob("*.csv")]                                                                                                   
    matcha_shops = [clean_shop_name(p) for p in MATCHA_DIR.glob("*.csv")]                                                                                                   
                                                                                                                                                                            
    # Format the list content                                                                                                                                               
    lines = ["Coffee Shops:"]                                                                                                                                               
    lines.extend(f"- {shop}" for shop in sorted(coffee_shops))                                                                                                              
    lines.extend(["", "Matcha Shops:"])                                                                                                                                     
    lines.extend(f"- {shop}" for shop in sorted(matcha_shops))                                                                                                              
                                                                                                                                                                            
    # Write to output file                                                                                                                                                  
    output_path = Path("../Dataset/shops_list.txt")                                                                                                                         
    output_path.write_text("\n".join(lines), encoding="utf-8")
```
