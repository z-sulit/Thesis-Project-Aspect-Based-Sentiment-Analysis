#Generate shops list

```python
import os                                                                                                                                                               
from pathlib import Path                                                                                                                                                
                                                                                                                                                                        
MATCHA_DIR = Path("../Dataset/Matcha Shops")                                                                                                                            
COFFEE_DIR = Path("../Dataset/Coffeeshops")                                                                                                                             
                                                                                                                                                                        
def clean_shop_name(path):                                                                                                                                              
    stem = path.stem                                                                                                                                                    
    if stem.endswith("_reviews"):                                                                                                                                       
        stem = stem[:-8]                                                                                                                                                
    return stem.replace("_", " ").strip()                                                                                                                               
                                                                                                                                                                        
# Collect and clean shop names from CSV files                                                                                                                           
coffee_shops = [clean_shop_name(p) for p in COFFEE_DIR.glob("*.csv")]                                                                                                   
matcha_shops = [clean_shop_name(p) for p in MATCHA_DIR.glob("*.csv")]                                                                                                   
                                                                                                                                                                        
# Format the list content                                                                                                                                               
lines = ["Coffee Shops:"]                                                                                                                                               
lines.extend(f"- {shop}" for shop in sorted(coffee_shops))                                                                                                              
lines.extend(["", "Matcha Shops:"])                                                                                                                                     
lines.extend(f"- {shop}" for shop in sorted(matcha_shops))                                                                                                              
                                                                                                                                                                        
# Write to output file                                                                                                                                                  
output_path = Path("../Dataset/shops_list.txt")                                                                                                                         
output_path.write_text("\n".join(lines), encoding="utf-8")
```

#Shop type column

```python
from pathlib import Path                                                                                                                                                                                     
import pandas as pd                                                                                                                                                                                          
                                                                                                                                                                                                                
# Check relative locations depending on where your Jupyter kernel is running                                                                                                                                 
if Path("../Dataset/Matcha Shops").exists():                                                                                                                                                                 
    MATCHA_DIR = Path("../Dataset/Matcha Shops")                                                                                                                                                             
    COFFEE_DIR = Path("../Dataset/Coffeeshops")                                                                                                                                                              
elif Path("Dataset/Matcha Shops").exists():                                                                                                                                                                  
    MATCHA_DIR = Path("Dataset/Matcha Shops")                                                                                                                                                                
    COFFEE_DIR = Path("Dataset/Coffeeshops")                                                                                                                                                                 
else:                                                                                                                                                                                                        
    # Fallback to absolute paths if relative paths cannot be found                                                                                                                                           
    MATCHA_DIR = Path(r"D:/Ateneo de Davao/(1) Thesis-Project-Aspect-Based-Sentiment-Analysis/Dataset/Matcha Shops")                                                                                         
    COFFEE_DIR = Path(r"D:/Ateneo de Davao/(1) Thesis-Project-Aspect-Based-Sentiment-Analysis/Dataset/Coffeeshops")                                                                                          
                                                                                                                                                                                                                
def add_shop_type_column(directory, shop_type):                                                                                                                                                              
    target_dir = Path(directory)                                                                                                                                                                             
    csv_files = list(target_dir.glob("*.csv"))                                                                                                                                                               
                                                                                                                                                                                                                
    if not csv_files:                                                                                                                                                                                        
        print(f"No CSV files found in: {target_dir.resolve()}")                                                                                                                                              
        return                                                                                                                                                                                               
                                                                                                                                                                                                                
    for csv_file in csv_files:                                                                                                                                                                               
        try:                                                                                                                                                                                                 
            df = pd.read_csv(csv_file)                                                                                                                                                                       
            df['Shop_Type'] = shop_type                                                                                                                                                                      
            df.to_csv(csv_file, index=False)                                                                                                                                                                 
            print(f"Successfully updated: {csv_file.name}")                                                                                                                                                  
        except Exception as e:                                                                                                                                                                               
            print(f"Error processing {csv_file.name}: {e}")                                                                                                                                                  
                                                                                                                                                                                                                
# Add shop type column to Matcha and Coffee CSV files                                                                                                                                                        
add_shop_type_column(MATCHA_DIR, "Matcha")                                                                                                                                                                   
add_shop_type_column(COFFEE_DIR, "Coffee")   
```
