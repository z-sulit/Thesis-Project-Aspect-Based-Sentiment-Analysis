#Generate shops list

```python
import os                                                                                                                                                               
from pathlib import Path                                                                                                                                                
                                                                                                                                                                        
MATCHA_DIR = Path("../Dataset/Matcha Shops")                                                                                                                            
COFFEE_DIR = Path("../Dataset/Coffeeshops")                                                                                                                             
                                                                                                                                                                        
def clean_shop_name(path):                                                                                                                                              
    stem = path.stem                                                                                                                                                    
    if stem.endswith("_reviews"):                                                                                                                                       
        stem = stem[:-8]                                                                                                                                                
    return stem.replace("_", " ").strip()                                                                                                                               
                                                                                                                                                                        
# Collect and clean shop names from CSV files                                                                                                                           
coffee_shops = [clean_shop_name(p) for p in COFFEE_DIR.glob("*.csv")]                                                                                                   
matcha_shops = [clean_shop_name(p) for p in MATCHA_DIR.glob("*.csv")]                                                                                                   
                                                                                                                                                                        
# Format the list content                                                                                                                                               
lines = ["Coffee Shops:"]                                                                                                                                               
lines.extend(f"- {shop}" for shop in sorted(coffee_shops))                                                                                                              
lines.extend(["", "Matcha Shops:"])                                                                                                                                     
lines.extend(f"- {shop}" for shop in sorted(matcha_shops))                                                                                                              
                                                                                                                                                                        
# Write to output file                                                                                                                                                  
output_path = Path("../Dataset/shops_list.txt")                                                                                                                         
output_path.write_text("\n".join(lines), encoding="utf-8")
```

#Shop type column

```python
from pathlib import Path                                                                                                                                                                                     
import pandas as pd                                                                                                                                                                                          
                                                                                                                                                                                                                
# Check relative locations depending on where your Jupyter kernel is running                                                                                                                                 
if Path("../Dataset/Matcha Shops").exists():                                                                                                                                                                 
    MATCHA_DIR = Path("../Dataset/Matcha Shops")                                                                                                                                                             
    COFFEE_DIR = Path("../Dataset/Coffeeshops")                                                                                                                                                              
elif Path("Dataset/Matcha Shops").exists():                                                                                                                                                                  
    MATCHA_DIR = Path("Dataset/Matcha Shops")                                                                                                                                                                
    COFFEE_DIR = Path("Dataset/Coffeeshops")                                                                                                                                                                 
else:                                                                                                                                                                                                        
    # Fallback to absolute paths if relative paths cannot be found                                                                                                                                           
    MATCHA_DIR = Path(r"D:/Ateneo de Davao/(1) Thesis-Project-Aspect-Based-Sentiment-Analysis/Dataset/Matcha Shops")                                                                                         
    COFFEE_DIR = Path(r"D:/Ateneo de Davao/(1) Thesis-Project-Aspect-Based-Sentiment-Analysis/Dataset/Coffeeshops")                                                                                          
                                                                                                                                                                                                                
def add_shop_type_column(directory, shop_type):                                                                                                                                                              
    target_dir = Path(directory)                                                                                                                                                                             
    csv_files = list(target_dir.glob("*.csv"))                                                                                                                                                               
                                                                                                                                                                                                                
    if not csv_files:                                                                                                                                                                                        
        print(f"No CSV files found in: {target_dir.resolve()}")                                                                                                                                              
        return                                                                                                                                                                                               
                                                                                                                                                                                                                
    for csv_file in csv_files:                                                                                                                                                                               
        try:                                                                                                                                                                                                 
            df = pd.read_csv(csv_file)                                                                                                                                                                       
            df['Shop_Type'] = shop_type                                                                                                                                                                      
            df.to_csv(csv_file, index=False)                                                                                                                                                                 
            print(f"Successfully updated: {csv_file.name}")                                                                                                                                                  
        except Exception as e:                                                                                                                                                                               
            print(f"Error processing {csv_file.name}: {e}")                                                                                                                                                  
                                                                                                                                                                                                                
# Add shop type column to Matcha and Coffee CSV files                                                                                                                                                        
add_shop_type_column(MATCHA_DIR, "Matcha")                                                                                                                                                                   
add_shop_type_column(COFFEE_DIR, "Coffee")   
```

#Concatenate CSVS

```python
from pathlib import Path                                                                                                                                               
import pandas as pd                                                                                                                                                    
                                                                                                                                                                        
# Hardcoded absolute path to Dataset directory                                                                                                                         
DATASET_DIR = Path(r"D:\Ateneo de Davao\Thesis-Project-Aspect-Based-Sentiment-Analysis\Dataset")                                                                       
                                                                                                                                                                        
coffee_dir = DATASET_DIR / "Coffeeshops"                                                                                                                               
matcha_dir = DATASET_DIR / "Matcha Shops"                                                                                                                              
                                                                                                                                                                        
coffee_out = DATASET_DIR / "coffeeshops_master.csv"                                                                                                                    
matcha_out = DATASET_DIR / "matcha_shops_master.csv"                                                                                                                   
                                                                                                                                                                        
                                                                                                                                                                        
def concatenate_folder(folder_path, default_shop_type, output_file):                                                                                                   
    csv_files = sorted([f for f in folder_path.glob("*.csv") if "master" not in f.name.lower()])                                                                       
    dfs = []                                                                                                                                                           
                                                                                                                                                                        
    for f in csv_files:                                                                                                                                                
        try:                                                                                                                                                           
            df = pd.read_csv(f, encoding="utf-8-sig")                                                                                                                  
        except Exception:                                                                                                                                              
            df = pd.read_csv(f, encoding="cp1252")                                                                                                                     
                                                                                                                                                                        
        if not df.empty:                                                                                                                                               
            if "Shop_Type" not in df.columns or df["Shop_Type"].isnull().all():                                                                                        
                df["Shop_Type"] = default_shop_type                                                                                                                    
            dfs.append(df)                                                                                                                                             
                                                                                                                                                                        
    if not dfs:                                                                                                                                                        
        print(f"No CSV files found in {folder_path}")                                                                                                                  
        return None                                                                                                                                                    
                                                                                                                                                                        
    master_df = pd.concat(dfs, ignore_index=True)                                                                                                                      
    master_df.to_csv(output_file, index=False, encoding="utf-8-sig")                                                                                                   
    print(f"SUCCESS: Saved {len(master_df):,} rows from {len(csv_files)} files to {output_file}")                                                                      
    return master_df                                                                                                                                                   
                                                                                                                                                                        
                                                                                                                                                                        
# Concatenate Coffeeshops and Matcha Shops                                                                                                                             
df_coffee = concatenate_folder(coffee_dir, "Coffee", coffee_out)                                                                                                       
df_matcha = concatenate_folder(matcha_dir, "Matcha", matcha_out)
```

```python
DATASET_DIR = Path(r"D:\Ateneo de Davao\Thesis-Project-Aspect-Based-Sentiment-Analysis\Dataset")                                                                       
                                                                                                                                                                        
coffee_file = DATASET_DIR / "coffeeshops_master.csv"                                                                                                                   
matcha_file = DATASET_DIR / "matcha_shops_master.csv" if (DATASET_DIR / "matcha_shops_master.csv").exists() else DATASET_DIR / "matcha_shops.csv"                      
output_file = DATASET_DIR / "master_reviews_combined.csv"                                                                                                              
                                                                                                                               
                                                                                                                                                                        
# Load Coffee Shops Master                                                                                                                                             
df_coffee = pd.read_csv(coffee_file, encoding="utf-8-sig")                                                                                                             
if "Shop_Type" not in df_coffee.columns or df_coffee["Shop_Type"].isnull().all():                                                                                      
    df_coffee["Shop_Type"] = "Coffee"                                                                                                                                  
                                                                                                                                                                        
# Load Matcha Shops Master                                                                                                                                             
df_matcha = pd.read_csv(matcha_file, encoding="utf-8-sig")                                                                                                             
if "Shop_Type" not in df_matcha.columns or df_matcha["Shop_Type"].isnull().all():                                                                                      
    df_matcha["Shop_Type"] = "Matcha"                                                                                                                                  
                                                                                                                                                                        
print(f"Loaded Coffee Shops Master: {len(df_coffee):,} rows")                                                                                                          
print(f"Loaded Matcha Shops Master: {len(df_matcha):,} rows")                                                                                                          
                                                                                                                                     
                                                                                                                                                                        
# Concatenate datasets                                                                                                                                                 
df_combined = pd.concat([df_coffee, df_matcha], ignore_index=True)                                                                                                     
                                                                                                                                                                        
# Save combined dataset to Dataset folder                                                                                                                              
df_combined.to_csv(output_file, index=False, encoding="utf-8-sig")                                                                                                     
                                                                                                                                                                        
print("=" * 60)                                                                                                                                                        
print(f"SUCCESS: Saved combined dataset to:\n  -> {output_file}")                                                                                                      
print(f"Total Rows: {len(df_combined):,} (Coffee: {len(df_coffee):,}, Matcha: {len(df_matcha):,})")                                                                    
print("=" * 60)                                                                                                                                                        
                                                                                                                                              
                                                                                                                                                                        
# Display dataset preview                                                                                                                                              
df_combined.head()     
```
