#Phase 4: Strict Data Partitioning

##Step 4.1: Pre-Augmentation Partitioning & Step 4.2: Evaluation Firewall

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split
```

```python
base_dir = 'D:/Ateneo de Davao/Thesis-Project-Aspect-Based-Sentiment-Analysis/_3_Granular Aspect Extraction and Data Decomposition'
file_path = os.path.join(base_dir, 'Exploded_Review_Aspect_Pairs.csv')

df = pd.read_csv(file_path)
df
```

```python
keep_cols = ['place_name', 'rating', 'shop_type', 'sentence', 'TargetAspect']
df_clean = df[keep_cols].copy()
```

```python
# Step 4.1: Pre-Augmentation Partitioning
# Split 1: 70% Train, 30% Temporary (to be split into Val/Test)
df_train, df_temp = train_test_split(
    df_clean, 
    test_size=0.30, 
    random_state=42, 
    stratify=df_clean['TargetAspect']
)

# Split 2: Divide the remaining 30% equally into 15% Validation and 15% Test
df_val, df_test = train_test_split(
    df_temp, 
    test_size=0.50, 
    random_state=42, 
    stratify=df_temp['TargetAspect']
)

# Lock files to establish the Evaluation Firewall
df_train.to_csv('Train_Set_70.csv', index=False, encoding='utf-8')
df_val.to_csv('Validation_Set_15.csv', index=False, encoding='utf-8')
df_test.to_csv('Test_Set_15.csv', index=False, encoding='utf-8')
```
