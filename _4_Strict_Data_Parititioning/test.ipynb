```python
import pandas as pd 
 
 
df_train = pd.read_csv('Augmented_Ground_Truth_714.csv') 
df_test = pd.read_csv('Test_Set_15.csv') 
 
 
 
train_sentences = set(df_train['sentence'].str.lower().str.strip()) 
test_sentences = set(df_test['sentence'].str.lower().str.strip()) 
 
 
shared = train_sentences.intersection(test_sentences)  
test_strings = df_test['sentence'].str.lower().str.strip() 
rows_affected = test_strings.isin(shared).sum() 
total_test_rows = len(df_test) 
leakage_pct = (rows_affected / total_test_rows) * 100 
print(f"Unique Overlapping Sentences: {len(shared)}") 
print(f"Total Test Rows Affected: {rows_affected} out of {total_test_rows}")
print(f"True Data Leakage Rate: {leakage_pct:.2f}%")
```
